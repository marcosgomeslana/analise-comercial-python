```python

import os
import pandas as pd


lista_arquivo = os.listdir("/content/drive/MyDrive/Vendas")

```

```python

table_total = pd.DataFrame()

for arquivo in lista_arquivo:

  if ("Vendas" in arquivo):
    table = pd.read_csv(f"/content/drive/MyDrive/Vendas/{arquivo}")
    table_total = pd.concat([table])




```

```python


display(table_total)
```

```python
#Produto mais vendido (em quantidade)

table_produto = table_total.groupby('Produto').sum()
table_produto = table_produto[['Quantidade Vendida']].sort_values(by='Quantidade Vendida', ascending=False)
display(table_produto)
```

```python
#Produto que mais faturou (em faturamento)

table_total['Faturamento'] = table_total['Quantidade Vendida'] * table_total['Preco Unitario']
table_faturamento = table_total.groupby('Produto').sum()
table_faturamento = table_faturamento[['Faturamento']].sort_values(by='Faturamento', ascending=False)
display(table_faturamento)
```

```python
#Loja ou cidade que mais vendeu (em faturamento) gráfico/dashboard

table_city = table_total.groupby('Loja').sum()
table_city = table_city[['Faturamento']].sort_values(by='Faturamento', ascending=False)
display(table_city)
```

```python
import plotly.express as px

grafico = px.bar(table_faturamento, x=table_faturamento.index, y='Faturamento')
grafico.show()
```
